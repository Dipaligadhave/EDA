```python
#find outliers
Q1=df['total_charges'].quantile(0.25)
Q3=df['total_charges'].quantile(0.75)

IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outlires=df[
    (df['total_charges']<lower)|
    (df['total_charges']>upper)
]
print(outlires)
```

```python
Q1 = df['monthly_charges'].quantile(0.25)
Q3 = df['monthly_charges'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['monthly_charges'] < lower) |
    (df['monthly_charges'] > upper)
]

print(outliers)
```
